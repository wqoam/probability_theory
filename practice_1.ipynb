### Задача 1.43
> Наудачу взятый телефонный номер состоит из 5 цифр.
Какова вероятность того, что в нем все цифры: а) различные;
б) одинаковые; в) нечетные? Известно, что номер телефона не начинается с цифры ноль.

#### a) - все цифры различные

```python
from itertools import product

nums = [x for x in range(0, 10)]

a = list(product(nums[1:], nums, nums, nums, nums))

k = 0
for line in a:
    if len(set(line)) == 5:
        k += 1

print(k / len(a))

```

#### б) - все цифры одинаковые

```python
k = 0
for line in a:
    if len(set(line)) == 1:
        k += 1

print(k / len(a))
```

#### в) - все цифры нечетные

```python
k = 0
for line in a:
    for x in line:
        if x % 2 == 0:
            break
    else:
        k += 1
# print(k)
print(f"{k / len(a):.4f}")
```

Найти вероятность того, что шар, вынутый из второй урны - белый.

```python
from random import sample
```

```python
def get_balls(urna : list, k : int):
    return sample(urna, k)

result = 0
N = 15_000_000
urna_1 = ["W"] * 15 + ["B"] * 5
for _ in range(N + 1):
    urna_2 = ["W"] * 6 + ["B"] * 1
    got_balls = get_balls(urna_1, 3)
    urna_2 += got_balls
    if get_balls(urna_2, 1) == ["W"]:
        result += 1

print(f"Результат: {result / N:.5f}")
```

### Задача №2.9)
> В среднем 20% пакетов акций на аукционах продаются
по первоначально заявленной цене. Найти вероятность того,
что из 9 пакетов акций в результате торгов по первоначально заявленной
цене: 
1. не будут проданы 5 пакетов; 
2. будет продано:
    * а) менее 2 пакетов; 
    * б) не более 2 пакетов; 
    * в) хотя бы 2 пакета;
    * г) наивероятнейшее число пакетов.

#### 1)

```python
# 1) 0.2 - по первоначальной
#    0.8 - не по первоначальной

n = 9
m = 5
p = 0.8
q = 1 - p

prob = stats.binom.pmf(m, n, p)
print(f"Вероятность того, что по первоначальной цене не будут проданы 5 пакетов: {prob:.4f}")
```

#### 2. a)

```python
# 2. a)

p = 0.2
q = 1 - p

prob_1 = 0

for m in range(0, 2):
    prob_1 += stats.binom.pmf(m, n, p)

print(f"Вероятность того, что по первоначальной цене будут проданы менее 2 пакетов: {prob_1:.4f}")
```

#### б)

```python
# 2. б)
m = 2

prob_2 = stats.binom.cdf(m, n, p)
print(f"Вероятность того, что по первоначальной цене будут проданы не более 2 пакетов: {prob_2:.4f}")
```

#### в)

```python
# 2. в)
m = 1 # хотя-бы 2 пакета = >1 пакета

prob_3 = stats.binom.sf(m, n, p)
print(f"Вероятность того, что по первоначальной цене будут проданы хотя-бы 2 пакета: {prob_3:.4f}")
```

#### г)

```python
# 2. г)

lowerb = n*p - q
upperb = n*p + p

most_n = [i for i in range(int(upperb) + 2) if lowerb <= i <= upperb]

prob_4 = 0

for m in most_n:
    prob_4 += stats.binom.pmf(m, n, p)

print(f"Вероятность того, что по первоначальной цене будут проданы наивероятнейшие число пакетов: {prob_4:.4f}")
```

1) будет повреждено: 
    * а) 3; 
    * б) по крайней мере 3; 
2) не будет повреждено:

```python
# 1. а) Будет повреждено 3 изделия
n = 10_000
m = 3
p = 0.02 / 100 # - вер-ть того, что изделие будет повреждено
q = 1 - p

lambd = n * p

prob = stats.poisson.pmf(m, lambd)
print(f"Вероятность того, что будет повреждено 3 детали из 10_000: {prob:.4f}")
```

#### б)

```python
# 1. б) Будет повреждено >=3 изделия
n = 10_000
m = 3
p = 0.02 / 100 # - вер-ть того, что изделие будет повреждено
q = 1 - p

lambd = n * p
prob = 1

for m in range(m):
    prob -= stats.poisson.pmf(m, lambd)
print(f"Вероятность того, что будет повреждено >=3 детали из 10_000: {prob:.4f}")
```

#### 2. a)

#### б)

```python
# 2. б) НЕ Будет повреждено хотя-бы 9997 изделия = будет повреждено <= 3 изделия
n = 10_000
m = 3
p = 0.02 / 100 # - вер-ть того, что изделие будет повреждено
q = 1 - p

lambd = n * p
prob = 0

for m in range(m + 1):
    prob += stats.poisson.pmf(m, lambd)
print(f"Вероятность того, что НЕ Будет повреждено хотя-бы 9997 изделия: {prob:.4f}")
```

### Задача №2.11

По результатам проверок налоговыми инспекциями
установлено, что в среднем каждое второе малое предприятие
региона имеет нарушение финансовой дисциплины. Найти
вероятность того, что из 1000 зарегистрированных в регионе малых
предприятий имеют нарушения финансовой дисциплины:
* а) 480 предприятий; 
* б) наивероятнейшее число предприятий; 
* в) не менее 480; 
* г) от 480 до 520.

#### а)

```python
# а) Имеют нарущения 480 из 1000

from math import sqrt

n = 1000
m = 480
p = 0.5
q = 1 - p

x = (m - n * p) / sqrt(n * p * q)
prob = stats.norm.pdf(x) / sqrt(n * p * q)

print(f"вероятность того, что 480 предприятий имеют нарушения: {prob:.4f}")
```

#### б)

```python
# б) Наивероятнейшее число предприятий.

lowerb = n*p - q
upperb = n*p + p

most_m = max([i for i in range(int(lowerb) - 1, int(upperb) + 2) if lowerb <= i <= upperb])

x = (most_m - n * p) / sqrt(n * p * q)
prob = stats.norm.pdf(x) / sqrt(n * p * q)

print(f"вероятность того, что наивероятнейшее число предприятий имеют нарушения: {prob:.4f}")
```

#### в)

```python
# в) Не менее 480 = >=480

a = 480
b = 1000
x1 = (a - n * p) / sqrt(n * p * q)
x2 = (b - n * p) / sqrt(n * p * q)

prob = stats.norm.cdf(x2) - stats.norm.cdf(x1)
print(f"вероятность того, что не менее 480 предприятий имеют нарушения: {prob:.4f}")
```

#### г)

```python
# г) от 480 до 520

a = 480
b = 520
x1 = (a - n * p) / sqrt(n * p * q)
x2 = (b - n * p) / sqrt(n * p * q)

prob = stats.norm.cdf(x2) - stats.norm.cdf(x1)
print(f"вероятность того, что от 480 до 520 предприятий имеют нарушения: {prob:.4f}")
```
