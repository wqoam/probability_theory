### Задача №1.8
> В конкурсе по 5 номинациям участвуют 10 кинофильмов. Сколько существует вариантов распределения призов, если по каждой номинации установлены: а) различные призы; б) одинаковые призы? 

#### a)

```python
# Пусть каждый кинофильм обозначается одной латинской буквой: 
items = "ABCDEFGHIG"
a = product(items, repeat=5)
print(f"Общее количество призов: {len(list(a))}")
```

#### б)

```python
a = list(combinations_with_replacement(items, r=5))
print(f"Общее количество призов: {len(list(a))}")
```

### Задача №1.10
> Буквы Т, Е, И, Я, Р, О написаны на отдельных
карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к друтой: а) 3 карточки; б) все 6 карточек. Какова вероятность того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»? 


#### a)


```python
words = "ТЕИЯРО"
a = list(permutations(words, 3))    # количество всевозможных комбинаций 3 букв из набора
prob = 1/len(a)                     # благоприятный исход / всевозможн комбинации
print(f"{prob:.4f}")
```

#### б)


```python
a = list(permutations(words, 6))
print(len(a))
prob = 1/len(a)
print(f"{prob:.4f}")
```

### Задача №1.11
> Используя условие примера 1.10, найти вероятность того, что получится слово «АНАНАС», если на отдельных
карточках написаны три буквы А, две буквы Н и одна буква С.

### Задача №1.16
> За круглым столом рассаживаются 5 мужчин и
5 женщин. Найти вероятность того, что: а) никакие два лица одного
пола не сядут рядом; б) супруги сядут рядом, если эти мужчины и
женщины образуют 5 супружеских пар. 

#### а)

```python
items = "МММММЖЖЖЖЖ"
a = list(permutations(items, r=10)) # кол-во способов рассадки 10 чел
print(len(a))

women = list(permutations("ЖЖЖЖЖ", r=5))
men = list(permutations("МММММ", r=5))

prob = (len(men) * len(women)) * 2 / len(a)
print(f"{prob:.4f}")
```

#### б)

```python
items = "ППППП"                     # П - пара
a = list(permutations(items, r=5))  # перестановка 5 пар за столом

items = "МММММЖЖЖЖЖ"
b = list(permutations(items, r=10)) # кол-во способов рассадки 10 чел

prob = len(a) * 2 ** 5 / len(b)     # 2**5 тк в паре М и Ж могут меняться местами, всего 5 пар
print(f"{prob:.4f}")



```

```python
passengers = "1234567"
places = "М" * 36 

a = len(list(combinations(places, r=7)))    # кол-во способов разместить 7 пассажир на 36 местах

b = len(list(combinations("K"*9, r=2)))

c = len(list(combinations("M"*8, r=7)))

prob = b * c / a

print(f"{prob:.7f}")
```
